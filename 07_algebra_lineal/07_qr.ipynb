    <span style="color:#FF0000"><b>Caso 1.</b></span> Dado el vector $X = \left( \begin{array}{c} 1 \\ 2 \\ 3 \\ 4 \end{array} \right) \in \mathbb{R}^4$, calcular la matriz de Householder $H(X) \in \mathcal{M}_4(\mathbb{R})$.
    </span></div>

```python
# Caso 1.
X = array([[1], [2], [3], [4]])
print("Vector: X = \n", X)

def householder(Z):
  m, n = shape(Z)
  return eye(m)-2/(conjugada(Z)@Z)*(Z@conjugada(Z))

print(householder(X))
```
