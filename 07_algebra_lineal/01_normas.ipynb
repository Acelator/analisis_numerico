<div align="left"><span style="font-family: Arial; color:#000000; font-size: medium">
    Las <b>variables</b> que utilizamos (elementos de los vectores y las matrices) pueden ser <b>enteras</b>, <b>reales</b> (caracterizadas por la presencia del punto decimal o por el uso de la notacion cientifica) o <b>complejas</b> (la unidad imaginaria se designa por <span style="font-family: Courier">j</span>, y debe ir precedida por la parte imaginaria del numero complejo sin espacio intermedio); el <b>tipo</b> de la variable se obtiene con la funcion <span style="font-family: Courier">type()</span>. Todo numero entero se puede convertir en real con la <b>funcion</b> <span style="font-family: Courier">float()</span>, mientras que todo numero entero o real se puede convertir en complejo con la <b>funcion</b> <span style="font-family: Courier">complex()</span>. Respecto a los numeros complejos estan disponibles, entre otras, las <b>funciones</b> <span style="font-family: Courier">real()</span>, <span style="font-family: Courier">imag()</span>, <span style="font-family: Courier">abs()</span>, <span style="font-family: Courier">angle()</span>, <span style="font-family: Courier">conj()</span> y <span style="font-family: Courier">exp()</span>, todas ellas con un significado obvio.
    </span></div>

```python
# Tipos de variables.
i = 4
print("Entero: ", i, " de tipo ", type(i))
print("Real: ", float(i), " de tipo ", type(float(i)))
print("Complejo: ", complex(i), " de tipo ", type(complex(i)))
```

```python
r1 = 5.
print("Real: ", r1, " de tipo ", type(r1))
print("Complejo: ", complex(r1), " de tipo ", type(complex(r1)))
```

```python
r2 = 6e0
print("Real: ", r2, " de tipo ", type(r2))
print("Complejo: ", complex(r2), " de tipo ", type(complex(r2)))
```

<div align="left"><span style="font-family: Arial; color:#000000; font-size: medium">
    Para definir un <b>vector</b> (fila) o una <b>matriz</b> utilizamos la <b>funcion</b> <span style="font-family: Courier">array()</span> cuyo <b>primer argumento</b> son los elementos del <span style="font-family: Courier">array</span>, separados por comas, encerrados entre corchetes y, en el caso de las matrices, las filas encerradas entre otros corchetes y tambien separadas por comas. La <b>funcion</b> <span style="font-family: Courier">print()</span> permite imprimirlos en pantalla. La <b>clase</b> a la que pertenecen (<span style="font-family: Courier">numpy.ndarray</span>) se obtiene con la <b>funcion</b> <span style="font-family: Courier">type()</span>, el <b>numero de dimensiones</b> (1 en el caso de vectores y 2 en el de matrices) con la <b>funcion</b> <span style="font-family: Courier">ndim()</span> (tambien se puede utilizar el <b>comando</b> <span style="font-family: Courier">.ndim</span>), el <b>tamano</b> (longitud del vector, o filas y columnas de las matrices) con la <b>funcion</b> <span style="font-family: Courier">shape()</span> (tambien se puede utilizar el <b>comando</b> <span style="font-family: Courier">.shape</span>), mientras que el <b>numero total de elementos</b> (longitud del vector, o producto del numero de filas por el de columnas de las matrices) con la <b>funcion</b> <span style="font-family: Courier">size()</span> (tambien se puede utilizar el <b>comando</b> <span style="font-family: Courier">.size</span>).
     </span></div>

```python
# Construccion de matrices.
A = array([[1, -2, 3], [-4, 5, -6]])
x = array([7., -8, 9])
X = array([[7, -8e0, 9]])
Y = array([[1j], [-2]])
```

```python
print("Impresion de las matrices.")
print("A = \n", A)
print("x = \n", x)
print("X = \n", X)
print("Y = \n", Y)
```

```python
print("Tipo de las matrices.")
print("type(A) = ", type(A))
print("type(x) = ", type(x))
print("type(X) = ", type(X))
print("type(Y) = ", type(Y))
```

```python
print("Numero de dimensiones de las matrices.")
print("ndim(A) = ", ndim(A))
print("ndim(x) = ", ndim(x))
print("ndim(X) = ", ndim(X))
print("ndim(Y) = ", ndim(Y))
print(A.ndim, x.ndim, X.ndim, Y.ndim)
```

```python
print("Tamano de las matrices.")
print("shape(A) = ", shape(A))
print("shape(x) = ", shape(x))
print("shape(X) = ", shape(X))
print("shape(Y) = ", shape(Y))
print(A.shape, x.shape, X.shape, Y.shape)
```

```python
print("Numero de elementos de las matrices.")
print("size(A) = ", size(A))
print("size(x) = ", size(x))
print("size(X) = ", size(X))
print("size(Y) = ", size(Y))
print(A.size, x.size, X.size, Y.size)
```

No obstante, se puede <i>forzar el tipo de los elementos</i> con el <b>argumento opcional</b> <span style="font-family: Courier">dtype=</span> de la funcion <span style="font-family: Courier">array()</span>.

<div align="left"><span style="font-family: Arial; color:#000000; font-size: medium">
    Se puede <b>crear</b> un <span style="font-family: Courier">array</span> lleno de 0's con la <b>funcion</b> <span style="font-family: Courier">zeros()</span> y lleno de 1's con la <b>funcion</b> <span style="font-family: Courier">ones()</span>; en ambos casos se debe pasar como <b>argumento</b> el tamano (<span style="font-family: Courier">shape</span>) del <span style="font-family: Courier">array</span>. Para generar un <span style="font-family: Courier">array</span> con <i>elementos aleatorios</i> siguiendo una distribucion uniforme sobre el intervalo [0, 1), la <b>funcion</b> es <span style="font-family: Courier">random.rand()</span>; esta funcion lleva tantos <b>argumentos</b> como dimensiones se quiere que tenga el <span style="font-family: Courier">array</span>, con los tamanos correspondientes. La <b>funcion</b> <span style="font-family: Courier">eye()</span> permite crear una matriz, no necesariamente cuadrada, cuyos elementos son todos 0, excepto los de la diagonal principal que valen 1; si esta funcion lleva un unico <b>argumento</b> se crea una matriz cuadrada de dicho tamano y si lleva dos una matriz rectangular; esta funcion admite un <b>argumento opcional</b> en la forma <span style="font-family: Courier">k=</span> que permite que la diagonal de 1's este <span style="font-family: Courier">k</span> posiciones por encima o por debajo (dependiendo del signo) de la diagonal principal.
    </span></div>

```python
# Matrices especiales.
print("Matriz de 0's.")
C = zeros((3, 2))
print("C = \n", C)
print("C.dtype = ", C.dtype)
```

```python
print("Matriz de 1's.")
D = ones((2, 3))
print("D = \n", D)
print("D.dtype = ", D.dtype)
```

```python
print("Matriz aleatoria.")
E = random.rand(3, 2)
print("E = \n", E)
print("E.dtype = ", E.dtype)
```

```python
print("Matriz de diagonal principal de 1's.")
F = eye(2, 3)
print("F = \n", F)
print("F.dtype = ", F.dtype)
```

```python
print("Matriz de k-diagonal principal de 1's.")
G = eye(3, k=1)
print("G = \n", G)
print("G.dtype = ", G.dtype)
```

<div align="left"><span style="font-family: Arial; color:#000000; font-size: medium">
    Las <b>funciones</b> <span style="font-family: Courier">arange()</span> y <span style="font-family: Courier">linspace()</span> permiten <b>crear vectores</b> (<span style="font-family: Courier">array</span> de una dimension) <i>interpolando</i> valores intermedios entre dos extremos; los tres <b>argumentos</b> de la funcion <span style="font-family: Courier">arange</span> son el extremo inicial del intervalo (cerrado), el extremo final (abierto) y el paso de malla; los tres <b>argumentos</b> de la funcion <span style="font-family: Courier">linspace</span> son el extremo inicial del intervalo (cerrado), el extremo final (cerrado) y el numero total de puntos interpolados. Se pueden <b>crear matrices</b> (<span style="font-family: Courier">array</span> de dos dimensiones) combinando estas funciones con la <b>funcion</b> <span style="font-family: Courier">reshape()</span>, cuyo primer <b>argumento</b> es un vector y el segundo el tamano de la matriz que se creara.
    </span></div>

```python
print("Matriz creada por interpolacion.")
h = arange(0, 12, 1)
print("h = ", h)
H = reshape(h, (3, 4))
print("H = \n", H)
print("H.dtype = ", H.dtype)
```

```python
print("Matriz creada por interpolacion.")
k = linspace(0, 11, 12)
print("k = ", k)
K = reshape(k, (4, 3))
print("K = \n", K)
print("K.dtype = ", K.dtype)
```

En el caso de matrices, la presencia de esta notacion en uno de los dos indices <b>genera un vector</b>, mientras que la presencia en ambos indices <b>genera una matriz</b> (submatriz de la matriz dada).

<div align="left"><span style="font-family: Arial; color:#000000; font-size: medium">

<div align="left"><span style="font-family: Arial; color:#000000; font-size: medium">

Cuando esta funcion tienen como argumento un vector el efecto es el mismo que la funcion anterior.
    </span></div>

```python
print("Matrices diagonales.")
print("diagflat(x) = \n", diagflat(x))
```

```python
print("diagflat(Y, k=1) = \n", diagflat(Y, k=1))
```

```python
print("diagflat(A, k=-1) = \n", diagflat(A, k=-1))
```

```python
print("diag(A) = ", diag(A))
```

```python
print("diag(A, k=1) = ", diag(A, k=1))
```

```python
print("diag(Y) = ", diag(Y))
```

```python
print("diag(x) = \n", diag(x))
```

  Caso 2.</b></span> Dada la matriz $A = \left( \begin{array}{cc} 1 & 2 \\ 3 & 4 \end{array} \right)$, construir las matrices $B = \left( \begin{array}{c|c} A & 2\,A \\ \hline 3\,A & 4\,A \end{array} \right)$ y $C = \left( \begin{array}{c|c|c} A & I & \\ \hline I & A & I \\ \hline & I & A \end{array} \right)$. Con la ayuda de la funcion <span style="font-family: Courier">det()</span>, cuyo unico argumento es una matriz cuadrada, calcular los determinantes de las matrices anteriores.
    </span></div>

```python
# Caso 2.
print("Caso 2.")

# Dada A simetrica
A = array([[1,2], [3,4]])
print("Det A = ", det(A))
B = vstack([hstack([A, 2*A]), hstack([3*A, 4*A])])
print("B = \n", B)
print("Det B = ", det(B))
C = vstack([hstack([A, eye(2), zeros((2, 2))]), hstack([eye(2), A, eye(2)]), hstack([zeros((2, 2)), eye(2), A])])
print("C = \n", C)
print("Det C = ", det(C))
```

  Caso 3.</b></span> Construir una matriz $A=(a_{i,j})\in\mathcal{M}_{10}(\mathbb{R})$ tal que $a_{i,j}=10\,(i-1)+j$, para todo $i,j=1,2,\ldots,10$. Construir tambien las matrices $B,C\in\mathcal{M}_{10}(\mathbb{R})$ que coinciden con $A$, excepto que en $B$ permutan entre si las filas y columnas segunda y quinta, y en $C$ a la quinta fila de $A$ se le suma 3 veces la segunda fila.
    </span></div>

```python
# Caso 3.
print("Caso 3.")
A = reshape(arange(1, 101, 1, dtype=float), (10, 10))
print("A = \n", A)
B = array(A)
B[[1, 4], :] = B[[4, 1], :]
B[:, [1, 4]] = B[:, [4, 1]]
print("B = \n", B)
C = array(A)
C[4, :] += 3*C[1, :]
print("C = \n", C)
```

  Se llama <b>matriz de Hilbert de orden $n$</b> a la matriz $H=(h_{i,j})\in\mathcal{M}_n(\mathbb{R})$ tal que $h_{i,j}=\frac{1}{i+j-1}$, para todo $i,j=1,2,\ldots,n$. Construir e imprimir una matriz de Hilbert de orden 5, asi como todas sus submatrices principales (que son tambien matrices de Hilbert); calcular tambien el determinante de estas submatrices.
    </span></div>

```python
# Caso 4.
print("Caso 4.")
n = 5
H = zeros((n,n))
for i in range(n):
    for j in range(n):
        H[i, j] = 1/(i + j + 1)
print("H = \n", H)
for k in range(n):
    print("Submatriz principal de orden: ", k+1)
    print("H^(", k+1, ") = \n", H[:k+1, :k+1])
    print("det H^(", k+1, ") = ",  det(H[:k+1, :k+1]))
```

    <span style="color:#FF0000"><b>Caso 5.</b></span>
  Se llama <b>matriz de Vandermonde de orden $n$</b> a la matriz $V=(v_{i,j})\in\mathcal{M}_n(\mathbb{K})$ tal que $v_{i,j}={\alpha_i}^{j-1}$, para todo $i,j=1,2,\ldots,n$, siendo $\alpha_i\in\mathbb{K}$, para $i=1,2,\ldots,n$; en el caso en que $\alpha_i=i$, construir e imprimir una matriz de Vandermonde de orden 5.
    <br>
  Para suprimir lineas (filas o columnas) de una matriz, se puede utilizar la funcion <span style="font-family: Courier">delete()</span>, cuyos tres argumentos son una matriz, la(s) linea(s) que se quieren suprimir, y <span style="font-family: Courier">axis=</span> para indicar si se suprimen filas o columnas. Por otro lado, la funcion <span style="font-family: Courier">inv()</span>, cuyo unico argumento es una matriz cuadrada, calcula la inversa de la misma (supuesta regular).
    <br>
  Construir la matriz de cofactores de la matriz de Vandermonde anterior y comprobar que la

transpuesta de la misma, dividida por su determinante, es la inversa.
</span>

```python
# Caso 5.
print("Caso 5.")
n = 5
V = zeros((n,n))
for i in range(n):
    for j in range(n):
        V[i, j] = (i + 1)**j
print("V = \n", V)
print("det(V) = ", det(V))
cofV = zeros((n,n))
for i in range(n):
    for j in range(n):
        cofV[i, j] = (-1)**(i + j)*det(delete(delete(V, i, axis=0), j, axis=1))
print("cof(V) = \n", cofV)
print("V^(-1) = \n", inv(V))
print("det(V)^(-1)*cof(V)^t = \n", transpose(cofV)/det(V))
print("Comprobacion: ", max(abs(inv(V) - transpose(cofV)/det(V))))
```
