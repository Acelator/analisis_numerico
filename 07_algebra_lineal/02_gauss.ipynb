    El primer objetivo de esta seccion es ver la forma en la que podemos <b>definir nuestras propias funciones</b> que implementen un determinado algoritmo.
    </span></div>

<div align="left"><span style="font-family: Arial; color:#000000; font-size: medium">
    Por ejemplo, nos podemos plantear definir una funcion, a la que llamaremos <span style="font-family: Courier">conjugada()</span>, que calcule la <b>conjugada</b> de una matriz.
    <br>
  Esta funcion debe tener un unico argumento de entrada, que es una matriz $A\in\mathcal{M}_{m\times n}(\mathbb{C})$, y debe devolver su matriz conjugada $A^*\in\mathcal{M}_{n\times m}(\mathbb{C})$. El codigo necesario para llevar a cabo esta tarea puede ser tan simple como este que sigue.
    </span></div>

```python
def conjugada(A):
    return conjugate(transpose(A))
```

<div align="left"><span style="font-family: Arial; color:#000000; font-size: medium">
    El siguiente objetivo de esta seccion es calcular los <b>autovalores</b> y <b>autovectores</b> de una matriz, <b>diagonalizarla</b> (si es posible) y calcular sus <b>valores singulares</b>.
    </span></div>

<div align="left"><span style="font-family: Arial; color:#000000; font-size: medium">
  Dada una matriz $A\in\mathcal{M}_n(\mathbb{K})$, sus <b>autovalores</b> y <b>autovectores</b> pueden calcularse con la ayuda de la funcion <span style="font-family: Courier">eig()</span>; esta funcion tiene un unico argumento de entrada que es la matriz cuadrada $A$ y tiene dos argumentos de salida, siendo el primero un vector de $n$ componentes conteniendo a los autovalores (en un orden arbitrario, quizas numeros complejos aunque la matriz sea de coeficientes reales y repetidos en caso de multiplicidad mayor que uno) y el segundo una matriz cuadrada de orden $n$ cuyas columnas son autovectores (asociados a los autovalores calculados, en el mismo orden y unitarios respecto del producto escalar ordinario).
    <br>
  No obstante la matriz de autovectores puede no ser regular en el caso de que la matriz dada no sea diagonalizable (los autovectores asociados a autovalores con multiplicidad mayor que 1 no son linealmente independientes).
    <br>
  En el caso de que la matriz dada si sea diagonalizable, la matriz de autovectores que obtenemos es regular, pero aunque tengamos asegurado que existe una matriz de paso ortogonal-unitaria (al ser la matriz dada normal), la que obtenemos no lo es necesariamente.
    </span></div>

<div align="left"><span style="font-family: Arial; color:#000000; font-size: medium">

    <span style="color:#FF0000"><b>Caso 3 (a).</b></span>
  Calcular los autovalores y autovectores, y ver si es diagonalizable la matriz $A = \left( \begin{array}{rr} 3 & 1 \\ 0 & 1 \end{array} \right)$.
    <br>
  Imprimir el vector de autovalores y la matriz de autovectores (ortogonal-unitaria si es posible) y comprobar si la matriz dada es diagonalizable (comprobar que $P^{-1}\,A\,P=D$ para lo cual se puede utilizar la funcion <span style="font-family: Courier">inv()</span> que calcula la inversa de la funcion dada como argumento).
    </span></div>

```python
# Caso 3-a.
print("Caso 3-a.")
A = array([[3, 1], [0, 1]])
print("Matriz: A = \n", A)
d, P = eig(A)
print("Vector de autovalores: d = ", d)
print("Matriz de autovectores: P = \n", P)
detP = det(P)
print("det(P) = ", detP)
if abs(detP) > 1e-10:
    print("La matriz es diagonalizable.")
    D = inv(P)@A@P
    print("inv(P)@A@P = \n", D)
    print("Comprobacion: ", max(abs(D - diag(d))))
else:
    print("La matriz no es diagonalizable.")
```

```python
# Matriz normal.
normal = A@conjugada(A) - conjugada(A)@A
print("A@A^* - A^*@A = \n", normal)
mv = max(abs(normal))
if mv > 1e-10:
    print("La matriz no es normal.")
else:
    print("La matriz es normal.")
```

```python
# Matriz de paso ortogonal-unitaria.
n = size(d)
unit = conjugada(P)@P
print("P^*@P = \n", unit)
mv = max(abs(unit - eye(n)))
if mv > 1e-10:
    print("La matriz de paso no es ortogonal-unitaria.")
else:
    print("La matriz de paso es ortogonal-unitaria.")
```

```python
# Autovalores y autovectores.
n = size(d)
for i in range(n):
    print("Autovalor: lambda = ", d[i], " Autovector (fila): x = ", P[:, i])
```

<div align="left"><span style="font-family: Arial; color:#000000; font-size: medium">
    Dada una matriz $A\in\mathcal{M}_n(\mathbb{K})$, sus <b>valores singulares</b> pueden calcularse con la ayuda de la funcion <span style="font-family: Courier">svd()</span>; esta funcion tiene un unico argumento de entrada que es la matriz cuadrada $A$ y tiene tres argumentos de salida, siendo el segundo un vector de $n$ componentes (reales y mayores o iguales que 0) conteniendo a los valores singulares (en un orden arbitrario y repetidos en caso de multiplicidad mayor que uno) y el primero y tercero las  dos matrices de orden $n$ ortogonales-unitarias de su descomposicion en valores singulares.
    <br>
    Si llamamos $S$ a la matriz diagonal cuyos elementos diagonales son los valores singulares de $A$ (en el orden que proporciona la funcion <span style="font-family: Courier">svd()</span>) y $U$ y $V$ las matrices ortogonales-unitarias que se obtienen como primera y tercera salida de dicha funcion, se verifica que $U\,S\,V=A \iff U^{-1}\,A\,V^{-1}=S$.
    </span></div>

<div align="left"><span style="font-family: Arial; color:#000000; font-size: medium">
    <span style="color:#FF0000"><b>Caso 4 (a).</b></span> Realizar la descomposicion en valores singulares de la matriz $A = \left( \begin{array}{rr} 3 & 1 \\ 0 & 1 \end{array} \right)$.
    </span></div>

```python
# Caso 4-a.
print("Caso 4-a.")
A = array([[3, 1], [0, 1]])
print("Matriz: A = \n", A)
U, s, V = svd(A)
print("Vector de valores singulares: s = ", s)
print("Matrices ortogonales-unitarias de la descomposicion en valores singulares:")
print("U = \n", U)
print("V = \n", V)
DVS = U@diag(s)@V
print("U@diag(s)@V = \n", DVS)
print("Comprobacion: ", max(abs(DVS - A)))
DVS1 = conjugada(U)@A@conjugada(V)
print("U^{-1}@A@V^{-1} = \n", DVS1)
print("Comprobacion: ", max(abs(DVS1 - diag(s))))
```

```python
# Caso 6. Caso del caso 3-f.
print("Caso 6 - 3-f.")
A = ones((4, 4)) + 6*eye(4)
print("Matriz: A = \n", A)
d, P = eig(A)
print("Vector de autovalores: d = ", d)
print("Matriz de autovectores: P = \n", P)
detP = det(P)
print("det(P) = ", detP)
if abs(detP) > 1e-10:
    print("La matriz es diagonalizable.")
    D = inv(P)@A@P
    print("inv(P)@A@P = \n", D)
    print("Comprobacion: ", max(abs(D - diag(d))))
else:
    print("La matriz no es diagonalizable.")
```
